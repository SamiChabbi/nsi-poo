> Écrire une fonction `comparer(carte1, carte2)` qui prend en paramètres deux objets de la classe `Carte`. Cette fonction renvoie :

>

#### ✅ RÉSOLUTION

````python
class Joueur:
    def __init__(self, pseudo, identifiant, equipe):
        self.pseudo = pseudo
        self.equipe = equipe
        self.id = identifiant
        self.nb_de_tirs_emis = 0
        self.liste_id_tirs_recus = []
        self.est_actif = True

    def tire(self):
        if self.est_actif == True:
            self.nb_de_tirs_emis = self.nb_de_tirs_emis + 1

    def est_determine(self):
        return self.nb_de_tirs_emis > 500

    def subit_un_tir(self, id_recu):
        if self.est_actif == True:
            self.est_actif = False
            self.liste_id_tirs_recus.append(id_recu)

    def redevenir_actif(self):
        if self.est_actif == False:
            self.est_actif = True

    def nb_de_tirs_recus(self):
        return len(self.liste_id_tirs_recus)
````

### Question 4

Recopier et compléter, en utilisant les méthodes des classes `Joueur` et `Base`, les 2 lignes de codes suivantes qu'il faut ajouter à la fin de la méthode `collecte_information` :

> *Dans le notebook, la classe `Base` est écrite pour vous : il reste les 2 lignes à compléter.*

#### ✅ RÉSOLUTION

````python
class Base:
    def __init__(self, equipe, liste_id):
        self.equipe = equipe
        self.liste_des_id_de_l_equipe = liste_id
        self.score = 1000

    def est_un_id_allie(self, id):
        return id in self.liste_des_id_de_l_equipe

    def incremente_score(self, n):
        self.score = self.score + n

    def collecte_information(self, participant):
        if participant.equipe == self.equipe:        # test 1
            for id in participant.liste_id_tirs_recus:
                if self.est_un_id_allie(id):          # test 2
                    self.incremente_score(-20)
                else:
                    self.incremente_score(-10)
            if participant.est_determine():
                self.incremente_score(40)
````

*Les tests de la feuille d'exercices :*

````python
base_a = Base("A", [1, 2, 3])
j1 = Joueur("Sniper", 1, "A")
for k in range(501):
    j1.tire()
j1.subit_un_tir(2)            # tir d'un coéquipier : -20
base_a.collecte_information(j1)
assert base_a.score == 1000 - 20 + 40
j2 = Joueur("Lent", 3, "A")
j2.subit_un_tir(9)            # tir ennemi : -10
base_a.collecte_information(j2)
assert base_a.score == 1020 - 10
print("✔ Tous les tests passent")
````

## Exercice 3 — L'agence immobilière  ⭐

Une agence immobilière développe un programme pour gérer les biens immobiliers qu'elle propose à la vente. Dans ce programme, pour modéliser les données de biens immobiliers, on définit une classe `Bim` avec les attributs suivants :

### Question 1

````python
class Bim:
    def __init__(self, nature, surface, prix_moy):
        self.nt = nature
        self.sf = surface
        self.pm = prix_moy

    def estim_prix(self):
        return self.sf * self.pm
````

### Question 3

#### 📝 ÉNONCÉ

> On souhaite affiner l'estimation du prix d'un bien en prenant en compte sa nature :
> - pour un bien dont l'attribut `nt` est `'maison'` la nouvelle estimation du prix est le produit de sa surface par le prix moyen par m² multiplié par 1,1 ;
> - pour un bien dont l'attribut `nt` est `'bureau'` la nouvelle estimation du prix est le produit de sa surface par le prix moyen par m² multiplié par 0,8 ;
> - pour les biens d'autres natures, l'estimation du prix ne change pas.
>
> Modifier le code de la méthode `estim_prix` afin de prendre en compte ce changement de calcul.

#### ✅ RÉSOLUTION

````python
class Bim:
    def __init__(self, nature, surface, prix_moy):
        self.nt = nature
        self.sf = surface
        self.pm = prix_moy

    def estim_prix(self):
        prix = self.sf * self.pm
        if self.nt == 'maison':
            return prix * 1.1
        elif self.nt == 'bureau':
            return prix * 0.8
        else:
            return prix
````

*Les tests de la feuille d'exercices :*

````python
assert abs(Bim('maison', 70.0, 2000.0).estim_prix() - 154000.0) < 0.01
assert abs(Bim('bureau', 100.0, 1000.0).estim_prix() - 80000.0) < 0.01
assert Bim('appartement', 50.0, 3000.0).estim_prix() == 150000.0
print("✔ Tous les tests passent")
````

#### 💡 EXPLICATION PÉDAGOGIQUE

- En Python, la virgule des nombres décimaux s'écrit avec un **point** : `1.1` et `0.8`.
- On calcule une fois le prix de base dans une variable, puis on le multiplie selon la nature : c'est plus lisible que de recopier trois fois `self.sf * self.pm`.

### Question 4

#### 📝 ÉNONCÉ

> Écrire le code Python d'une fonction `nb_maison(lst)` qui prend en argument une liste Python de biens immobiliers de type `Bim` et qui renvoie le nombre d'objets de nature `'maison'` contenus dans la liste `lst`.

#### ✅ RÉSOLUTION

````python
class Compte:
    def __init__(self, numero, adherent, adresse):
        self.numero = numero
        self.adherent = adherent
        self.adresse = adresse
        self.solde = 0

    def crediter(self, montant):
        self.solde = self.solde + montant

    def debiter(self, montant):
        self.solde = self.solde - montant

    def est_positif(self):
        return self.solde >= 0

    def transferer(self, autre_compte, montant):
        self.debiter(montant)
        autre_compte.crediter(montant)
````

> Pour en faciliter la gestion, on crée une liste Python `liste_comptes` contenant tous les comptes des adhérents. Le gestionnaire de l'association souhaite relancer les adhérents dont le compte est débiteur, c'est-à-dire dont le solde est négatif.
> 1. Écrire la fonction `rechercher_debiteurs` qui prend en argument `liste_comptes` et renvoie une liste de dictionnaires dont la première clé est le nom de l'adhérent et la seconde clé le solde de son compte en négatif. Exemple de résultat :
> `liste_debiteurs = [{'Nom': 'DUPONT Thomas', 'solde': -60}, {'Nom': 'CARNEIRO Sarah', 'solde': -75}]`
> 2. Écrire la fonction `urgent_debiteur` qui prend en argument la liste de dictionnaires et renvoie le nom de l'adhérent le plus endetté. On suppose qu'aucun adhérent n'a la même dette.

#### ✅ RÉSOLUTION

````python
def rechercher_debiteurs(liste_comptes):
    debiteurs = []
    for c in liste_comptes:
        if not c.est_positif():
            debiteurs.append({'Nom': c.adherent, 'solde': c.solde})
    return debiteurs


def urgent_debiteur(liste_debiteurs):
    pire = liste_debiteurs[0]
    for d in liste_debiteurs:
        if d['solde'] < pire['solde']:
            pire = d
    return pire['Nom']
````

#### ✅ RÉSOLUTION

````python
import random


class Personnage:
    def __init__(self, coordx, coordy, coordz, point_de_vie, point_degats):
        self.x = coordx
        self.y = coordy
        self.z = coordz
        self.vie = point_de_vie
        self.degats = point_degats

    def avancex(self):
        self.x = self.x + 1

    def raz(self):
        self.x = 0
        self.y = 0
        self.z = 0

    def coord(self):
        return (self.x, self.y, self.z)

    def get_etat(self):
        return self.vie

    def newgame(self):
        if self.vie <= 0:
            self.raz()
            self.vie = 15

    def potionmystere(self):
        if random.randint(1, 2) == 1:
            nbPoint = -1
        else:
            nbPoint = +1
        self.vie = self.vie + nbPoint

    def piege(self):
        self.vie = self.vie - 10

    def repos(self):
        self.vie = self.vie + 5

    def perdre_vie(self, points):
        self.vie = self.vie - points
        self.newgame()

    def attaquer(self, autre):
        autre.perdre_vie(self.degats)
````

````python
class Date:
    def __init__(self, jour, mois, annee):
        self.jour = jour
        self.mois = mois
        self.annee = annee
        self.nb_jours_par_mois = [31, 28, 31, 30, 31, 30, 31, 31, 30, 31, 30, 31]
        if self.est_bissextile():
            self.nb_jours_par_mois[1] = 29

    def get_jour(self):
        return self.jour

    def get_mois(self):
        return self.mois

    def get_annee(self):
        return self.annee

    def set_jour(self, jour):
        self.jour = jour

    def set_mois(self, mois):
        self.mois = mois

    def set_annee(self, annee):
        self.annee = annee

    def est_bissextile(self):
        a = self.annee
        return (a % 4 == 0 and a % 100 != 0) or a % 400 == 0
````

````python
class Region:
    def __init__(self, nom_region):
        self.nom = nom_region
        self.tab_voisines = []
        self.tab_couleurs_disponibles = ['rouge', 'vert', 'bleu', 'jaune', 'orange', 'marron']
        self.couleur_attribuee = None

    def renvoie_premiere_couleur_disponible(self):
        return self.tab_couleurs_disponibles[0]

    def renvoie_nb_voisines(self):
        return len(self.tab_voisines)

    def est_coloriee(self):
        return self.couleur_attribuee != None

    def retire_couleur(self, couleur):
        if couleur in self.tab_couleurs_disponibles:
            self.tab_couleurs_disponibles.remove(couleur)

    def est_voisine(self, region):
        for voisine in self.tab_voisines:
            if voisine == region:
                return True
        return False
````

#### 📝 ÉNONCÉ

> Le code Python ci-après présente la fonction récursive `chargement_glouton` dont les paramètres sont :

> - `rang` : un indice compris entre 0 inclus et `len(liste)` inclus ;
> - `capacite` : une charge exprimée en kilogrammes ;
>
> et qui renvoie la liste des colis à charger en appliquant l'algorithme glouton, en supposant que la charge restante dans le camion est `capacite`, et en ne considérant que les colis de `liste` d'indice supérieur ou égal à `rang`.
> ```python
> 1  def chargement_glouton(liste, rang, capacite):
> 2      if rang == len(liste):
> 3          return ...
> 4      elif liste[rang].poids <= ...:
> 5          return ... + chargement_glouton(liste, ..., ...)
> 6      else:
> 7          return chargement_glouton(liste, ..., ...)
> ```
> Recopier et compléter le code de la fonction `chargement_glouton`.

#### ✅ RÉSOLUTION

````python
def chargement_glouton(liste, rang, capacite):
    if rang == len(liste):
        return []
    elif liste[rang].poids <= capacite:
        return [liste[rang]] + chargement_glouton(liste, rang + 1, capacite - liste[rang].poids)
    else:
        return chargement_glouton(liste, rang + 1, capacite)
````

  - **le colis tient** : on le prend (`[liste[rang]]`, une liste d'un seul élément) et on continue avec le colis suivant et une **capacité diminuée** de son poids ;
  - **le colis ne tient pas** : on passe au suivant, la capacité ne change pas.
- `[liste[rang]] + ...` colle deux listes : il faut des crochets autour de `liste[rang]`.

On suppose que la liste est non vide. Recopier et compléter le code Python, donné ci-dessous, de la fonction `meilleur_concurrent`.
> ```python
> 1   def meilleur_concurrent(L):
> 2       conc_mini = L. ...
> 3       mini = conc_mini.temps_tot
> 4       Q = L.queue()
> 5       while not (Q.est_vide()):
> 6           elt = Q.tete()

> 11      return ...

#### ✅ RÉSOLUTION

````python
def meilleur_concurrent(L):
    conc_mini = L.tete()
    mini = conc_mini.temps_tot
    Q = L.queue()
    while not (Q.est_vide()):
        elt = Q.tete()
        if elt.temps_tot < mini:
            conc_mini = elt
            mini = elt.temps_tot
        Q = Q.queue()
    return conc_mini
````

### Partie A — Questions 1, 2 et 3

#### 📝 ÉNONCÉ

Le constructeur de cette classe prend en paramètre une valeur (de type `int`) comprise entre 1 et 104 et l'affecte à l'attribut `valeur` (de type `int`). On ajoute l'attribut `TdB` (de type `int`) qui contient le nombre de « têtes de bœuf » qui sera calculé à l'aide de la méthode `calcul_TdB` (définie à la question suivante) qui n'a pas de paramètre.

> Chaque carte possède une ou plusieurs « têtes de bœuf ». Ce nombre est déterminé de la façon suivante :
> - si la valeur de la carte est divisible par 11 alors la carte reçoit 5 « têtes de bœuf » ;
> - si la valeur de la carte se termine par 0 alors la carte reçoit 3 « têtes de bœuf » ;
> - si la valeur de la carte se termine par 5 alors la carte reçoit 2 « têtes de bœuf » ;
> - si la valeur de la carte ne remplit aucune de ces conditions alors elle ne reçoit qu'1 « tête de bœuf ».
>
> Attention, ces règles se cumulent. Par exemple, 55 est divisible par 11 et se termine par 5 donc la carte 55 comporte 7 « têtes de bœuf ».
>
> 2. Écrire le code de la méthode `calcul_TdB` de la classe `Carte` afin de calculer le nombre de « têtes de bœuf » pour une carte donnée.

#### ✅ RÉSOLUTION

````python
class Carte:
    def __init__(self, valeur):
        self.valeur = valeur
        self.TdB = self.calcul_TdB()

    def calcul_TdB(self):
        tdb = 0
        if self.valeur % 11 == 0:
            tdb = tdb + 5
        if self.valeur % 10 == 0:
            tdb = tdb + 3
        if self.valeur % 10 == 5:
            tdb = tdb + 2
        if tdb == 0:
            tdb = 1
        return tdb

    def est_superieure_a(self, autre):
        return self.valeur > autre.valeur
````

#### ✅ RÉSOLUTION

**b.** La région renvoyée :

#### 💡 EXPLICATION PÉDAGOGIQUE

- C'est le modèle 5, **le meilleur**, avec un départ un peu différent : au lieu de partir du premier élément, on part de `None` et de −1 (toute région a au moins 0 voisine, donc la première région rencontrée remplace forcément −1).

**Code donné par le sujet** (exécuter la cellule). *La dernière ligne sert seulement dans le notebook : elle ajoute la méthode à la classe `Pays` écrite plus haut.*

````python
def renvoie_max(self):
    nb_voisines_max = -1
    region_max = None
    for reg in self.renvoie_tab_regions_non_coloriees():
        if reg.renvoie_nb_voisines() > nb_voisines_max:
            nb_voisines_max = reg.renvoie_nb_voisines()
            region_max = reg
    return region_max


Pays.renvoie_max = renvoie_max   # ajoute cette méthode à la classe Pays déjà écrite
````

### Question 11

#### 📝 ÉNONCÉ

> Coder la méthode `colorie(self)` de la classe `Pays` qui choisit une couleur pour chaque région du pays de la façon suivante :
> - On récupère la région non coloriée qui possède le plus de voisines.
> - Tant que cette région existe :
>   - La couleur attribuée à cette région est la première couleur disponible dans son tableau de couleurs disponibles.
>   - Pour chaque région voisine de la région : si la couleur choisie est présente dans le tableau des couleurs disponibles de la région voisine alors on la retire de ce tableau.
>   - On récupère à nouveau la région non coloriée qui possède le plus de voisines.

#### ✅ RÉSOLUTION

````python
def colorie(self):
    region = self.renvoie_max()
    while region != None:
        couleur = region.renvoie_premiere_couleur_disponible()
        region.couleur_attribuee = couleur
        for voisine in region.tab_voisines:
            voisine.retire_couleur(couleur)
        region = self.renvoie_max()


Pays.colorie = colorie   # ajoute la méthode à la classe (ligne à garder)
````
